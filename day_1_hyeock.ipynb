```python
from pathlib import Path

import h5py
import kagglehub

# Download the dataset from Kaggle instead of reading files from ./archive.
dataset_path = Path(
    kagglehub.dataset_download("itshpark/data-driven-prediction-of-battery-cycle")
)
print("Path to dataset files:", dataset_path)

mat_files = sorted(dataset_path.rglob("*.mat"))
if not mat_files:
    raise FileNotFoundError(f"No .mat files found in {dataset_path}")

print("MAT files:", [file.name for file in mat_files])
battery_data = {file.stem: h5py.File(file, "r") for file in mat_files}
print(f"Opened {len(battery_data)} MAT files")
print("Variables in first file:", list(next(iter(battery_data.values())).keys()))
```
